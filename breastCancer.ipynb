```python
import pandas as pd
from skimage.transform import resize
from skimage.io import imread
import os
from keras.models import Sequential
from keras.layers import  Conv2D,Dense,MaxPool2D,Flatten
```

```python
from google.colab import drive
drive.mount('/content/drive')
```

```python
datadir='/content/drive/MyDrive/breast cancer/ultrasound breast classification/train'
os.listdir(datadir)
```

```python
X_train=[]
Y_train=[]
categories=['malignant', 'benign']
for category in categories:
  path=os.path.join(datadir,category)
  images=os.listdir(path)
  for img in images:
    imgpath=os.path.join(path,img)
    img_array=imread(imgpath)
    img_resized=resize(img_array,(150,150,1))
    X_train.append(img_resized)
    Y_train.append(categories.index(category))
  print("Loading.....",category)
```

```python
import seaborn as sns
import pandas as pd

df = pd.DataFrame(Y_train)
df.value_counts()
```

```python
datadir='/content/drive/MyDrive/breast cancer/ultrasound breast classification/val'
X_test=[]
Y_test=[]
categories=['malignant', 'benign']
for category in categories:
  path=os.path.join(datadir,category)
  images=os.listdir(path)
  for img in images:
    imgpath=os.path.join(path,img)
    img_array=imread(imgpath)
    img_resized=resize(img_array,(150,150,1))
    X_test.append(img_resized)
    Y_test.append(categories.index(category))
  print("Loading.....",category)
```

```python
import numpy as np
X_train=np.array(X_train)
X_test=np.array(X_test)
Y_train=np.array(Y_train)
Y_test=np.array(Y_test)
```

```python
X_train.shape,Y_train.shape
```

```python
X_test.shape,Y_test.shape
```

```python
model=Sequential()
model.add(Conv2D(32,(3,3),input_shape=(150,150,1),activation='relu'))
model.add(MaxPool2D(2,2))
model.add(Conv2D(32,(3,3),activation='relu'))
model.add(MaxPool2D(2,2))
model.add(Flatten())
model.add(Dense(100,activation='relu'))
model.add(Dense(2,activation='sigmoid'))
```

```python
model.compile(optimizer='adam',loss='sparse_categorical_crossentropy',metrics=['accuracy'])
```

```python
model.fit(X_train,Y_train,epochs=50,batch_size=12)
```

```python
model.evaluate(X_test,Y_test)
```
