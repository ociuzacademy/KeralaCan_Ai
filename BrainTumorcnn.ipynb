```python
import pandas as pd
from skimage.transform import resize     #to resize and normalize image
from skimage.io import imread   #read image
import os #
from keras.models import Sequential
from keras.layers import  Conv2D,Dense,MaxPool2D,Flatten
```

```python
datadir='/content/drive/MyDrive/Brain Tumor/Training-20241009T063458Z-001/Training'
os.listdir(datadir)
```

```python
X_tarin=[]
Y_train=[]
categories=['pituitary', 'meningioma', 'glioma', 'notumor']
for category in categories:
  path=os.path.join(datadir,category)
  images=os.listdir(path)
  for img in images:
    imgpath=os.path.join(path,img)
    img_array=imread(imgpath)
    img_resized=resize(img_array,(150,150,1))
    X_tarin.append(img_resized)
    Y_train.append(categories.index(category))
  print("Loading.....",category)
```

```python
datadir='/content/drive/MyDrive/Brain Tumor/Testing-20241009T063506Z-001/Testing'
X_test=[]
Y_test=[]
categories=['pituitary', 'meningioma', 'glioma', 'notumor']
for category in categories:
  path=os.path.join(datadir,category)
  images=os.listdir(path)
  for img in images:
    imgpath=os.path.join(path,img)
    img_array=imread(imgpath)
    img_resized=resize(img_array,(150,150,1))
    X_test.append(img_resized)
    Y_test.append(categories.index(category))
  print("Loading.....",category)
```

```python
import numpy as np
X_tarin=np.array(X_tarin)
X_test=np.array(X_test)
Y_train=np.array(Y_train)
Y_test=np.array(Y_test)
```

```python
X_tarin.shape,Y_train.shape
```

```python
X_test.shape,Y_test.shape
```

```python
model=Sequential()
model.add(Conv2D(32,(3,3),input_shape=(150,150,1),activation='relu'))
model.add(MaxPool2D(2,2))
model.add(Conv2D(32,(3,3),activation='relu'))
model.add(MaxPool2D(2,2))
model.add(Flatten())
model.add(Dense(100,activation='relu'))
model.add(Dense(4,activation='softmax'))
```

```python
model.compile(optimizer='adam',loss='sparse_categorical_crossentropy',metrics=['accuracy'])
```

```python
model.fit(X_tarin,Y_train,epochs=20,batch_size=12)
```

```python
model.evaluate(X_test,Y_test)
```
